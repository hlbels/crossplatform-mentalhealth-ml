```python
# import libraries
import re
import os
import pandas as pd
import joblib
from sklearn.feature_extraction.text import TfidfVectorizer, ENGLISH_STOP_WORDS
from sklearn.model_selection import train_test_split
```

```python
# functions
# Text Cleaning
def clean_text(t):
    # URL removal
    t = re.sub(r"http\S+", "", t)
    # special character removal
    t = re.sub(r"[^a-zA-Z\s]", "", t)
    # whitespace normalization
    t = re.sub(r"\s+", " ", t).strip()
    return t

# Tokenization and Normalization
def tokenize_normalize(t, model):
    # For logistinc, SVM and RF, apply tokenization, lowercasing and stopword handling
    if model == "tfidf":
        t = t.lower()
        tokens = t.split()
        tokens = [w for w in tokens if w not in ENGLISH_STOP_WORDS]
        return " ".join(tokens)
    # for BiLSTM, apply tokenization and lowercasing
    if model == "bilstm":
        t = t.lower()
        tokens = t.split()
        return " ".join(tokens)
    # for BERT, apply the tokenizer from the library later

# Feature Representation
def feature_representation(t):
    vectorizer = TfidfVectorizer(max_features=5000)
    X = vectorizer.fit_transform(t)
    return X, vectorizer

# Train/Validation/Test Split
def split_data(df):
    train, tmp = train_test_split(df, test_size=0.3, random_state=42)
    val, test = train_test_split(tmp, test_size=0.5, random_state=42)
    return train, val, test
```

```python
# reddit dataset
# mental health
reddit_mh_path = "data/raw/reddit/mental_health"
reddit_mh_files = [f for f in os.listdir(reddit_mh_path)]
mh_train = []
mh_val = []
mh_test = []

for f in reddit_mh_files:
    file_path = os.path.join(reddit_mh_path, f)
    df = pd.read_csv(file_path)
    # pick up useful keys
    df = df[['subreddit', 'post']]
    # add label: mental health = 1, control = 0
    df['label'] = 1
    df['cleaned_text'] = df['post'].apply(clean_text)
    df["tfidf_text"] = df["cleaned_text"].apply(lambda x: tokenize_normalize(x, model="tfidf"))
    df["bilstm_text"] = df["cleaned_text"].apply(lambda x: tokenize_normalize(x, model="bilstm"))
    # split each subreddit(category) into 70% train, 15% validation and 15% test
    train, val, test = split_data(df)
    mh_train.append(train)
    mh_val.append(val)
    mh_test.append(test)


# control
reddit_c_path = "data/raw/reddit/control"
reddit_c_files = [f for f in os.listdir(reddit_c_path)]
c_train = []
c_val = []
c_test = []

for f in reddit_c_files:
    file_path = os.path.join(reddit_c_path, f)
    df = pd.read_csv(file_path)
    # pick up useful keys
    df = df[['subreddit', 'post']]
    # add label: mental health = 1, control = 0
    df['label'] = 0
    df['cleaned_text'] = df['post'].apply(clean_text)
    df["tfidf_text"] = df["cleaned_text"].apply(lambda x: tokenize_normalize(x, model="tfidf"))
    df["bilstm_text"] = df["cleaned_text"].apply(lambda x: tokenize_normalize(x, model="bilstm"))
    # split each subreddit(category) into 70% train, 15% validation and 15% test
    train, val, test = split_data(df)
    c_train.append(train)
    c_val.append(val)
    c_test.append(test)


# combine mental health and control data
df_reddit_train = pd.concat(mh_train+ c_train, ignore_index=True)
df_reddit_val = pd.concat(mh_val + c_val, ignore_index=True)
df_reddit_test = pd.concat(mh_test + c_val, ignore_index=True)


# ↓↓ This part has too many steps, but it works for now... ↓↓
# Split each subset (train, val, test) into inputs directly usable by each model

# training data
# for logistinc, SVM and RF
X_tfidf, vectorizer = feature_representation(df_reddit_train["tfidf_text"])
joblib.dump(vectorizer, "data/preprocessed/reddit/vectorizer/reddit_train.pkl")
df_tfidf = pd.DataFrame(X_tfidf.toarray())
df_tfidf["subreddit"] = df_reddit_train["subreddit"]
df_tfidf["label"] = df_reddit_train["label"]
df_tfidf.to_csv("data/preprocessed/reddit/reddit_tfidf_train.csv", index=False)
# for BilSTM
df_bilstm = df_reddit_train[["bilstm_text", "subreddit", "label"]]
df_bilstm.to_csv("data/preprocessed/reddit/reddit_bilstm_train.csv", index=False)
# for BERT
df_bert = df_reddit_train[["cleaned_text", "subreddit", "label"]]
df_bert.to_csv("data/preprocessed/reddit/reddit_bert_train.csv", index=False)

# val data
# for logistic, SVM and RF
X_tfidf, vectorizer = feature_representation(df_reddit_val["tfidf_text"])
joblib.dump(vectorizer, "data/preprocessed/reddit/vectorizer/reddit_val.pkl")
df_tfidf = pd.DataFrame(X_tfidf.toarray())
df_tfidf["subreddit"] = df_reddit_val["subreddit"]
df_tfidf["label"] = df_reddit_val["label"]
df_tfidf.to_csv("data/preprocessed/reddit/reddit_tfidf_val.csv", index=False)
# for BilSTM
df_bilstm = df_reddit_val[["bilstm_text", "subreddit", "label"]]
df_bilstm.to_csv("data/preprocessed/reddit/reddit_bilstm_val.csv", index=False)
# for BERT
df_bert = df_reddit_val[["cleaned_text", "subreddit", "label"]]
df_bert.to_csv("data/preprocessed/reddit/reddit_bert_val.csv", index=False)

# test data
# for logistic, SVM and RF
X_tfidf, vectorizer = feature_representation(df_reddit_test["tfidf_text"])
joblib.dump(vectorizer, "data/preprocessed/reddit/vectorizer/reddit_test.pkl")
df_tfidf = pd.DataFrame(X_tfidf.toarray())
df_tfidf["subreddit"] = df_reddit_test["subreddit"]
df_tfidf["label"] = df_reddit_test["label"]
df_tfidf.to_csv("data/preprocessed/reddit/reddit_tfidf_test.csv", index=False)
# for BilSTM
df_bilstm = df_reddit_test[["bilstm_text", "subreddit", "label"]]
df_bilstm.to_csv("data/preprocessed/reddit/reddit_bilstm_test.csv", index=False)
# for BERT
df_bert = df_reddit_test[["cleaned_text", "subreddit", "label"]]
df_bert.to_csv("data/preprocessed/reddit/reddit_bert_test.csv", index=False)
```
