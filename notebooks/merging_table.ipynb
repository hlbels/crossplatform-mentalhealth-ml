```python
"""
Generate TF-IDF Visualization for Reddit Subreddits
Replaces the 12 individual tables with a professional visualization
"""

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Set style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

# Data from your paper - top 5 terms for each subreddit
tfidf_data = {
    'Addiction': [
        ('addict', 0.214672),
        ('drug', 0.076950),
        ('day', 0.067528),
        ('get', 0.065629),
        ('use', 0.061273)
    ],
    'ADHD': [
        ('adhd', 0.171457),
        ('get', 0.071245),
        ('take', 0.063267),
        ('like', 0.062595),
        ('work', 0.061374)
    ],
    'Alcoholism': [
        ('drink', 0.256658),
        ('alcohol', 0.167938),
        ('day', 0.074951),
        ('get', 0.056835),
        ('go', 0.056533)
    ],
    'Anxiety': [
        ('anxieti', 0.159875),
        ('feel', 0.088627),
        ('like', 0.071982),
        ('get', 0.071200),
        ('go', 0.061709)
    ],
    'Mental Health': [
        ('feel', 0.096978),
        ('like', 0.078385),
        ('mental', 0.064661),
        ('get', 0.063512),
        ('know', 0.063400)
    ],
    'SuicideWatch': [
        ('want', 0.094379),
        ('life', 0.077753),
        ('suicid', 0.074028),
        ('feel', 0.073488),
        ('kill', 0.066454)
    ],
    'Conspiracy': [
        ('peopl', 0.060316),
        ('us', 0.058049),
        ('amp', 0.054204),
        ('post', 0.046448),
        ('think', 0.045261)
    ],
    'Fitness': [
        ('weight', 0.140917),
        ('day', 0.080484),
        ('week', 0.070322),
        ('work', 0.068492),
        ('bodi', 0.065078)
    ],
    'Jokes': [
        ('whi', 0.067538),
        ('say', 0.061524),
        ('call', 0.050155),
        ('ask', 0.042655),
        ('one', 0.039385)
    ],
    'Meditation': [
        ('mind', 0.089553),
        ('feel', 0.067988),
        ('thought', 0.065265),
        ('like', 0.062536),
        ('ani', 0.058849)
    ],
    'Relationship': [
        ('relationship', 0.106324),
        ('want', 0.100882),
        ('friend', 0.098368),
        ('like', 0.085623),
        ('feel', 0.079081)
    ],
    'Teaching': [
        ('school', 0.150161),
        ('ani', 0.076035),
        ('work', 0.068330),
        ('year', 0.065723),
        ('would', 0.065253)
    ]
}

# Categories
mental_health_categories = ['Addiction', 'ADHD', 'Alcoholism', 'Anxiety', 'Mental Health', 'SuicideWatch']
control_categories = ['Conspiracy', 'Fitness', 'Jokes', 'Meditation', 'Relationship', 'Teaching']

# ============================================================================
# VISUALIZATION 1: Grouped Bar Chart (Top Term per Subreddit)
# ============================================================================

def create_top_terms_comparison():
    """Create bar chart showing top TF-IDF term for each subreddit"""

    subreddits = list(tfidf_data.keys())
    top_terms = [data[0][0] for data in tfidf_data.values()]
    top_scores = [data[0][1] for data in tfidf_data.values()]

    # Colors: mental health in red tones, control in blue tones
    colors = ['#d62728' if sub in mental_health_categories else '#1f77b4'
              for sub in subreddits]

    fig, ax = plt.subplots(figsize=(14, 6))

    bars = ax.bar(range(len(subreddits)), top_scores, color=colors, alpha=0.7, edgecolor='black')

    ax.set_xlabel('Subreddit', fontsize=12, fontweight='bold')
    ax.set_ylabel('TF-IDF Score', fontsize=12, fontweight='bold')
    ax.set_title('Top TF-IDF Term by Subreddit Category', fontsize=14, fontweight='bold', pad=20)
    ax.set_xticks(range(len(subreddits)))
    ax.set_xticklabels(subreddits, rotation=45, ha='right')
    ax.grid(True, alpha=0.3, axis='y')

    # Add term labels on bars
    for i, (bar, term) in enumerate(zip(bars, top_terms)):
        height = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2., height + 0.005,
                f'"{term}"', ha='center', va='bottom', fontsize=9, fontweight='bold')

    # Add legend
    from matplotlib.patches import Patch
    legend_elements = [
        Patch(facecolor='#d62728', alpha=0.7, edgecolor='black', label='Mental Health'),
        Patch(facecolor='#1f77b4', alpha=0.7, edgecolor='black', label='Control')
    ]
    ax.legend(handles=legend_elements, loc='upper right', fontsize=11)

    plt.tight_layout()
    plt.savefig('../data/processed/reddit_tfidf_top_terms.png', dpi=300, bbox_inches='tight')
    print("✓ Created: reddit_tfidf_top_terms.png")
    plt.close()

# ============================================================================
# VISUALIZATION 2: Heatmap of Top 5 Terms per Subreddit
# ============================================================================

def create_tfidf_heatmap():
    """Create heatmap showing top 5 terms for each subreddit"""

    # Prepare data for heatmap
    all_subreddits = list(tfidf_data.keys())

    # Create matrix: rows = subreddits, columns = term positions (1-5)
    data_matrix = []
    term_labels = []

    for subreddit in all_subreddits:
        scores = [term[1] for term in tfidf_data[subreddit]]
        data_matrix.append(scores)
        terms = [term[0] for term in tfidf_data[subreddit]]
        term_labels.append(terms)

    data_matrix = np.array(data_matrix)

    # Create figure
    fig, ax = plt.subplots(figsize=(10, 12))

    # Create heatmap
    im = ax.imshow(data_matrix, cmap='YlOrRd', aspect='auto')

    # Set ticks and labels
    ax.set_xticks(np.arange(5))
    ax.set_yticks(np.arange(len(all_subreddits)))
    ax.set_xticklabels(['1st', '2nd', '3rd', '4th', '5th'], fontsize=11)
    ax.set_yticklabels(all_subreddits, fontsize=11)

    ax.set_xlabel('Top Term Rank', fontsize=12, fontweight='bold')
    ax.set_ylabel('Subreddit', fontsize=12, fontweight='bold')
    ax.set_title('TF-IDF Scores: Top 5 Terms per Subreddit', fontsize=14, fontweight='bold', pad=20)

    # Add text annotations with term names and scores
    for i in range(len(all_subreddits)):
        for j in range(5):
            term = term_labels[i][j]
            score = data_matrix[i, j]
            text = ax.text(j, i, f'{term}\n{score:.3f}',
                          ha="center", va="center", color="black", fontsize=8)

    # Add colorbar
    cbar = plt.colorbar(im, ax=ax)
    cbar.set_label('TF-IDF Score', rotation=270, labelpad=20, fontsize=11, fontweight='bold')

    # Add category separators
    ax.axhline(y=5.5, color='blue', linewidth=2, linestyle='--', alpha=0.5)
    ax.text(5.2, 2.5, 'Mental\nHealth', fontsize=10, fontweight='bold',
            rotation=270, va='center', color='red')
    ax.text(5.2, 8.5, 'Control', fontsize=10, fontweight='bold',
            rotation=270, va='center', color='blue')

    plt.tight_layout()
    plt.savefig('../data/processed/reddit_tfidf_heatmap.png', dpi=300, bbox_inches='tight')
    print("✓ Created: reddit_tfidf_heatmap.png")
    plt.close()

# ============================================================================
# VISUALIZATION 3: Comparison - Mental Health vs Control
# ============================================================================

def create_category_comparison():
    """Create side-by-side comparison of mental health vs control subreddits"""

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))

    # Mental Health Categories
    mh_subreddits = mental_health_categories
    mh_data = {sub: tfidf_data[sub] for sub in mh_subreddits}

    for idx, (subreddit, terms) in enumerate(mh_data.items()):
        words = [t[0] for t in terms]
        scores = [t[1] for t in terms]
        y_pos = np.arange(len(words)) + idx * 6
        ax1.barh(y_pos, scores, height=0.8, label=subreddit, alpha=0.8)

        # Add labels
        for i, (word, score) in enumerate(zip(words, scores)):
            ax1.text(score + 0.005, y_pos[i], word, va='center', fontsize=8)

    ax1.set_xlabel('TF-IDF Score', fontsize=12, fontweight='bold')
    ax1.set_title('Mental Health Subreddits', fontsize=13, fontweight='bold')
    ax1.set_yticks([])
    ax1.legend(loc='lower right', fontsize=9)
    ax1.grid(True, alpha=0.3, axis='x')

    # Control Categories
    control_subs = control_categories
    control_data = {sub: tfidf_data[sub] for sub in control_subs}

    for idx, (subreddit, terms) in enumerate(control_data.items()):
        words = [t[0] for t in terms]
        scores = [t[1] for t in terms]
        y_pos = np.arange(len(words)) + idx * 6
        ax2.barh(y_pos, scores, height=0.8, label=subreddit, alpha=0.8)

        # Add labels
        for i, (word, score) in enumerate(zip(words, scores)):
            ax2.text(score + 0.005, y_pos[i], word, va='center', fontsize=8)

    ax2.set_xlabel('TF-IDF Score', fontsize=12, fontweight='bold')
    ax2.set_title('Control Subreddits', fontsize=13, fontweight='bold')
    ax2.set_yticks([])
    ax2.legend(loc='lower right', fontsize=9)
    ax2.grid(True, alpha=0.3, axis='x')

    plt.suptitle('TF-IDF Analysis: Mental Health vs Control Subreddits',
                 fontsize=15, fontweight='bold', y=0.98)

    plt.tight_layout()
    plt.savefig('../data/processed/reddit_tfidf_category_comparison.png', dpi=300, bbox_inches='tight')
    print("✓ Created: reddit_tfidf_category_comparison.png")
    plt.close()

# ============================================================================
# Main Execution
# ============================================================================

if __name__ == "__main__":
    print("="*60)
    print("GENERATING TF-IDF VISUALIZATIONS")
    print("="*60)

    create_top_terms_comparison()
    create_tfidf_heatmap()
    create_category_comparison()

    print("\n" + "="*60)
    print("ALL VISUALIZATIONS CREATED SUCCESSFULLY!")
    print("="*60)
    print("\nGenerated 3 figures:")
    print("1. reddit_tfidf_top_terms.png - Bar chart of top terms")
    print("2. reddit_tfidf_heatmap.png - Heatmap of all top 5 terms")
    print("3. reddit_tfidf_category_comparison.png - Mental Health vs Control")
```
